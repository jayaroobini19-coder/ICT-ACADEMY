#Part F – Data Analysis Using SQL


Answer the following questions using SQLite queries.



```python
#11 Display the Top 5 countries by GDP.
import sqlite3
conn = sqlite3.connect('conutry_statistics.db')
q = "SELECT country, GDP FROM country_statistics ORDER BY GDP DESC LIMIT 5"
top5_gdp = pd.read_sql_query(q, conn)
conn.close()
print(top5_gdp)
```

```python
#12 Find the average literacy rate
import sqlite3
conn = sqlite3.connect('conutry_statistics.db')
q = "SELECT AVG(literacy_rate) FROM country_statistics"
avg_literacy_rate = pd.read_sql_query(q, conn)
conn.close()
print(avg_literacy_rate)
```

```python
#13 Display all countries having literacy rate above 90%
conn = sqlite3.connect('conutry_statistics.db')
q = "SELECT country , literacy_rate  FROM country_statistics WHERE literacy_rate > 90 ORDER BY literacy_rate DESC"
df_filtered = pd.read_sql_query(q, conn)
conn.close()
print(df_filtered)
```

```python
# 14 Find the Top 5 countries with the highest internet penetration.
import sqlite3
conn = sqlite3.connect('conutry_statistics.db')
q = "SELECT country, \"internet_penetration (%)\" FROM country_statistics ORDER BY \"internet_penetration (%)\" DESC LIMIT 5"
top5_internet_penetration = pd.read_sql_query(q, conn)
conn.close()
print(top5_internet_penetration)
```

```python
#15 Calculate the total population.
import sqlite3
conn = sqlite3.connect('conutry_statistics.db')
q = "SELECT SUM(population) FROM country_statistics"
total_population = pd.read_sql_query(q, conn)
conn.close()
print(total_population)
```

```python
#16 Find the country with the highest GDP per capita.
import sqlite3
conn = sqlite3.connect('conutry_statistics.db')
q = 'SELECT country , gdp_Per_Capita FROM country_statistics  ORDER BY gdp_Per_Capita  DESC LIMIT 1'
df_filtered_gdp_Per_Capita = pd.read_sql_query(q,conn)
conn.close()
print(df_filtered_gdp_Per_Capita)
print("\n")
```

```python
#17 Export the final merged dataset as a CSV file
import sqlite3
conn = sqlite3.connect('conutry_statistics.db')
final_df = pd.read_sql_query("SELECT * FROM country_statistics", conn)
conn.close()
final_df.to_csv('country_statics.csv', index = False)
print("Dataset exported to 'country_statistics.csv',")
print("\n")
```

```python
#18 Sort the dataset in descending order of GDP per Capita and display the Top
import sqlite3
conn = sqlite3.connect('conutry_statistics.db')
q = "SELECT country, gdp_Per_Capita FROM country_statistics ORDER BY gdp_Per_Capita DESC LIMIT 10"
df_top10_gdp_Per_capita = pd.read_sql_query(q, conn)
conn.close()
print(df_top10_gdp_Per_capita)
print("\n")
```
